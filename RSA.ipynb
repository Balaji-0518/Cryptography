```python
from Crypto.PublicKey import RSA
p = 3
q = 17
n = p * q  # n = 3 * 11 = 33
e = 13
d = 5
m = 7      # Message, 0 <= m < n

# Validate parameters
phi_n = (p - 1) * (q - 1)  # phi(n) = 20
if m < 0 or m >= n:
    print(f"Error: m must be between 0 and {n-1}")
    exit()

# Create RSA key object (optional)
key = RSA.construct((n, e, d))
print(f"Public key: (e={e}, n={n})")
print(f"Private key: (d={d}, n={n})")

# Step 2: Encryption (c = m^e mod n)
c = pow(m, e, n)  # Raw RSA encryption
print(f"Ciphertext: c = {c}")

# Step 3: Decryption (m = c^d mod n)
decrypted_m = pow(c, d, n)  # Raw RSA decryption
print(f"Decrypted message: m = {decrypted_m}")

# Verify
if decrypted_m == m:
    print("Success: Decrypted message matches original m")
else:
    print("Error: Decryption failed")
```
